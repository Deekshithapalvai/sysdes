48. Recursive binary search	Implement binary search using recursion on a sorted list.

```python
def binary_search(numbers, target, left, right):
    if left > right:
        return -1
    mid = (left + right) // 2
    if numbers[mid] == target:
        return mid
    elif target < numbers[mid]:
        return binary_search(numbers, target, left, mid - 1)
    else:
        return binary_search(numbers, target, mid + 1, right)
numbers = [10, 20, 30, 40, 50, 60, 70]
target = 50
result = binary_search(numbers, target, 0, len(numbers) - 1)
print(result)
```

55. Bank account closure	Create create_bank_account() with nested functions for deposit, withdrawal, and balance checking.

```python
def create_bank_account(initial_balance):
    balance = initial_balance
    def deposit(amount):
        nonlocal balance
        if amount > 0:
            balance += amount
            return f"Deposited: {amount}"
        return "Invalid deposit amount"
    def withdraw(amount):
        nonlocal balance
        if amount <= 0:
            return "Invalid withdrawal amount"
        if amount > balance:
            return "Insufficient balance"
            balance -= amount
        return f"Withdrawn: {amount}"
    def check_balance():
        return f"Current balance: {balance}"
    return deposit, withdraw, check_balance
deposit, withdraw, check_balance = create_bank_account(1000)
print(check_balance())
print(deposit(500))
print(check_balance())
```

69	Memoization	Implement a memoization function that caches previous results of an expensive calculation.

```python
def memoize(function):
    cache = {}
    def wrapper(n):
        if n in cache:
            return cache[n]
        result = function(n)
        cache[n] = result
        return result
    return wrapper
@memoize
def square(n):
    print("Calculating...")
    return n * n
print(square(5))
print(square(5))
```

70	Data transformation pipeline	Build a pipeline of functions to clean transaction data, filter invalid records, calculate totals, and generate a summary. Each function should do one job.

```python
def clean_transactions(transactions: list[dict]) -> list[dict]:
    cleaned = []
    for transaction in transactions:
        record = transaction.copy()
        if isinstance(record.get("category"), str):
            record["category"] = record["category"].strip().title()
        cleaned.append(record)
    return cleaned
def filter_valid_transactions(transactions: list[dict]) -> list[dict]:
    valid = []
    for transaction in transactions:
        amount = transaction.get("amount")
        if (
            transaction.get("transaction_id")
            and isinstance(amount, (int, float))
            and not isinstance(amount, bool)
            and amount > 0
        ):
            valid.append(transaction)
    return valid
def calculate_totals(transactions: list[dict]) -> dict:
    totals = {}
    for transaction in transactions:
        category = transaction["category"]
        amount = transaction["amount"]
        totals[category] = totals.get(category, 0) + amount
    return totals
def generate_summary(transactions: list[dict]) -> dict:
    return {
        "total_transactions": len(transactions),
        "total_amount": sum(t["amount"] for t in transactions),
        "category_totals": calculate_totals(transactions)
    }
def transaction_pipeline(transactions: list[dict]) -> dict:
    cleaned = clean_transactions(transactions)
    valid = filter_valid_transactions(cleaned)
    summary = generate_summary(valid)
    return summary
transactions = [
    {"transaction_id": 101, "category": " grocery ", "amount": 500},
    {"transaction_id": 102, "category": "shopping", "amount": 1200},
    {"transaction_id": 103, "category": "grocery", "amount": 300},
    {"transaction_id": 104, "category": "food", "amount": -100},
    {"transaction_id": 105, "category": "food", "amount": 700}
]
print(transaction_pipeline(transactions))
```

```python
from functools import wraps
def log_execution(func):
    @wraps(func)
    def wrapper(*args, **kwargs):
        print("Function is executing")
        return func(*args, **kwargs)
    return wrapper
@log_execution
def calculate_total(a, b):
    """Calculate the sum of two numbers."""
    return a + b
print(calculate_total(10, 5))
print("\nFunction name:", calculate_total.__name__)
print("Docstring:", calculate_total.__doc__)
```


78	Retry decorator	Create a decorator that retries a failing function up to a configurable maximum number of attempts.

```python
def retry(max_attempts):
    def decorator(function):
        def wrapper():
            for attempt in range(max_attempts):
                try:
                    return function()
                except Exception:
                    print("Retrying...")
            return "All attempts failed"
        return wrapper
    return decorator
count = 0
@retry(3)
def process():
    global count
    count += 1
    if count < 3:
        raise ValueError("Error occurred")
    return "Success"
print(process())
```


79	Cache decorator	Build a decorator that caches the results of a function and avoids repeated calculations.


```python
def cache_result(func):
    cache = {}
    def wrapper(*args, **kwargs):
        key = (args, tuple(sorted(kwargs.items())))
        if key in cache:
            print("Returning cached result")
            return cache[key]
        print("Calculating result")
        result = func(*args, **kwargs)
        cache[key] = result
        return result
    return wrapper
@cache_result
def calculate_square(n):
    """Calculate the square of a number."""
    return n * n
print(calculate_square(5))
```

80	Class-based decorator	Create a class with __init__ and __call__ that works as a decorator and counts how many times a function is called.

```python
class CountCalls:
    def __init__(self, func):
        self.func = func
        self.count = 0
    def __call__(self, *args, **kwargs):
        self.count += 1
        print(f"Function called {self.count} times")
        return self.func(*args, **kwargs)
@CountCalls
def greet(name):
    """Greet a user."""
    return f"Hello, {name}!"
print(greet("Deekshitha"))
print(greet("Kaushik"))
print("Total calls:", greet.count)
```

```python
def data_cleaning(records: list[dict[str, str]]) -> list[dict[str, str]]:
    if not isinstance(records, list):
        raise TypeError("Input must be a list")
    cleaned_records = []
    for record in records:
        if not isinstance(record, dict):
            raise ValueError("Each record must be a dictionary")
        name = record.get("name")
        if not isinstance(name, str):
            continue
        name = name.strip().title()
        if not name:
            continue
        cleaned_records.append({
            "name": name
        })
    return cleaned_records
records = [
    {"name": "Deekshitha  "},
    {"name": "Deekshitha"},
    {"name": "  DEEKSHITHA"},
    {"name": ""},
    {"name": None}
]
print(data_cleaning(records))
```

99. Function-based ETL pipeline	Create separate functions for extracting CSV data, validating records, transforming data, and loading the output. Connect them into a reusable ETL pipeline.

```python
import csv
def extract_data(file_path):
    data = []
    with open(file_path, mode="r", newline="") as file:
        reader = csv.DictReader(file)
        for row in reader:
            data.append(row)
    return data
    
def validate_data(data):
    valid_records = []
    invalid_records = []
    for row in data:
        if (not row["EmployeeID"].strip()
            or not row["Name"].strip()
            or not row["Age"].strip()
            or not row["Salary"].strip()):
            invalid_records.append(row)
        else:
            valid_records.append(row)
    print("Valid records:", len(valid_records))
    print("Invalid records:", len(invalid_records))
    return valid_records

def transform_data(data):
    transformed_data = []
    for row in data:
        transformed_row = {
            "EmployeeID": int(row["EmployeeID"]),
            "Name": row["Name"].strip().title(),
            "Age": int(row["Age"]),
            "Salary": float(row["Salary"])
        }
        transformed_data.append(transformed_row)
    return transformed_data

def load_data(data, output_file):
    if not data:
        print("No data available to load.")
        return
    with open(output_file, mode="w", newline="") as file:
        fieldnames = data[0].keys()
        writer = csv.DictWriter(file, fieldnames=fieldnames)
        writer.writeheader()
        writer.writerows(data)
    print(f"Data successfully loaded into {output_file}")

def etl_pipeline(input_file, output_file):
    extracted_data = extract_data(input_file)
    valid_data = validate_data(extracted_data)
    transformed_data = transform_data(valid_data)
    load_data(transformed_data, output_file)
etl_pipeline("employees.csv", "employees_output.csv")
```
